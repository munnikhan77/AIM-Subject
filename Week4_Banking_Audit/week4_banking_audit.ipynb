```python


def clean_transactions(amounts, min_threshold=10.0):
    """Return only real money values that are positive and at least min_threshold."""
    is_valid = lambda x: (
        isinstance(x, (int, float))
        and not isinstance(x, bool)   
        and x > 0
        and x >= min_threshold
    )
    return list(filter(is_valid, amounts))


def apply_service_fee(amounts, fee_rate=0.02):
    """Return each amount after the fee is taken off, rounded to cents."""
    return list(map(lambda x: round(x - x * fee_rate, 2), amounts))


def calculate_nested_total(batch):
    """Add every number inside a list, no matter how deep the nesting goes."""
    if isinstance(batch, (int, float)):        
        return float(batch)
    if len(batch) == 0:                       
        return 0.0
    first, rest = batch[0], batch[1:]         
    return calculate_nested_total(first) + calculate_nested_total(rest)
```

```python

morning_batch = [250.75, -40, 0, 8.50, 10, 99.99, True, "75", 1200, 9.999]
cleaned = clean_transactions(morning_batch)
print("Raw batch      :", morning_batch)
print("Cleaned batch  :", cleaned)
print("After 2% fee   :", apply_service_fee(cleaned))
print("Strict (>=100) :", clean_transactions(morning_batch, 100))
print("1.5% fee rate  :", apply_service_fee(cleaned, 0.015))
```

```python

wire_transfer = [500, [120.25, 79.75], [[30, [20]], 50]]
print("\nNested batch   :", wire_transfer)
print("Nested total   :", calculate_nested_total(wire_transfer))
print("Empty batch    :", calculate_nested_total([]))
print("Single amount  :", calculate_nested_total(42))
print("Empty branches :", calculate_nested_total([[], [[]], 15]))
print("Deep (6 levels):", calculate_nested_total([1, [2, [3, [4, [5, [6.5]]]]]]))
```

```python

import sys
print("\nRecursion limit:", sys.getrecursionlimit())
try:
    calculate_nested_total(list(range(5000)))
except RecursionError as err:
    print("RecursionError on a 5,000-item flat batch:", err)
```
